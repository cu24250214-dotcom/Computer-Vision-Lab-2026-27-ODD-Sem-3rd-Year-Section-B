**Step 1-> Install and import the required libraries, including OpenCV, NumPy, and Matplotlib.**



```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
from google.colab import files

print("Upload your image:")
uploaded = files.upload()

file_name = list(uploaded.keys())[0]
img_bgr = cv2.imread(file_name)
```

**Step 2->Load a color image from the local system and display it using both OpenCV and Matplotlib.**

```python
img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(8, 4))
plt.subplot(1, 2, 1)
plt.imshow(img_bgr)
plt.title("Raw BGR (Inverted Colors)")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(img_rgb)
plt.title("Corrected RGB")
plt.axis("off")
plt.show()
```

**Step 3-> Examine the image properties such as dimensions, resolution, number of channels, and data type.**

```python
height, width, channels = img_bgr.shape
print(
    f"Height: {height}px | Width: {width}px | Channels: {channels} | Data Type: {img_bgr.dtype}"
)
```

**Step 4-> Save the image in different file formats (e.g., JPEG and PNG) and compare the outputs.**

```python
cv2.imwrite("compressed_output.jpg", img_bgr, [cv2.IMWRITE_JPEG_QUALITY, 50])

cv2.imwrite("lossless_output.png", img_bgr, [cv2.IMWRITE_PNG_COMPRESSION, 9])
```

**Step 5-> Convert the image from RGB/BGR to Grayscale, HSV, and LAB color spaces.**

```python
gray = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2GRAY)
hsv = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2HSV)
lab = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2LAB)

plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.imshow(gray, cmap="gray")
plt.title("Grayscale")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(hsv)
plt.title("HSV Color Space")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(lab)
plt.title("LAB Color Space")
plt.axis("off")

plt.tight_layout()
plt.show()
```

**Step 6->Perform basic geometric transformations such as resizing, rotation, and horizontal/vertical flipping.**

```python
resized = cv2.resize(img_bgr, (width // 2, height // 2))
flipped_h = cv2.flip(img_bgr, 1)

center = (width // 2, height // 2)
rot_matrix = cv2.getRotationMatrix2D(center, angle=45, scale=1.0)
rotated = cv2.warpAffine(img_bgr, rot_matrix, (width, height))

plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.imshow(cv2.cvtColor(resized, cv2.COLOR_BGR2RGB))
plt.title("Resized (50% Scale)")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(cv2.cvtColor(flipped_h, cv2.COLOR_BGR2RGB))
plt.title("Horizontal Flip")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(cv2.cvtColor(rotated, cv2.COLOR_BGR2RGB))
plt.title("Rotated 45°")
plt.axis("off")

plt.tight_layout()
plt.show()
```

**Step 7->Generate the complement (negative) of the image by inverting pixel intensity values.**

```python
img_negative = 255 - img_bgr

plt.figure(figsize=(8, 4))

plt.subplot(1, 2, 1)
plt.imshow(img_rgb)
plt.title("Original (RGB)")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(img_negative, cv2.COLOR_BGR2RGB))
plt.title("Negative Image")
plt.axis("off")

plt.tight_layout()
plt.show()
```

**Step 8->Crop a Region of Interest (ROI) from the image and analyze its characteristics.**

```python
roi = img_bgr[
    height // 4 : (3 * height) // 4, width // 4 : (3 * width) // 4
]

print(f"ROI Matrix Shape: {roi.shape}")
print(f"ROI Mean Intensity: {np.mean(roi):.2f}\n")

plt.figure(figsize=(8, 4))

plt.subplot(1, 2, 1)
plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(roi, cv2.COLOR_BGR2RGB))
plt.title("Cropped ROI (Central 50%)")
plt.axis("off")

plt.tight_layout()
plt.show()
```

**Step 9->Display the original and processed images for comparison and observe the effects of each operation.**

```python
titles = [
    "Original",
    "Grayscale",
    "HSV",
    "LAB",
    "Resized",
    "Rotated 45°",
    "Flipped",
    "Negative",
    "ROI",
]
processed_images = [
    img_rgb,
    gray,
    hsv,
    lab,
    cv2.cvtColor(resized, cv2.COLOR_BGR2RGB),
    cv2.cvtColor(rotated, cv2.COLOR_BGR2RGB),
    cv2.cvtColor(flipped_h, cv2.COLOR_BGR2RGB),
    cv2.cvtColor(img_negative, cv2.COLOR_BGR2RGB),
    cv2.cvtColor(roi, cv2.COLOR_BGR2RGB),
]

plt.figure(figsize=(10, 8))
for i in range(len(processed_images)):
    plt.subplot(3, 3, i + 1)
    if len(processed_images[i].shape) == 2:
        plt.imshow(processed_images[i], cmap="gray")
    else:
        plt.imshow(processed_images[i])
    plt.title(titles[i])
    plt.axis("off")

plt.tight_layout()
plt.show()
```

## Questions and Answers

1. What is a digital image? Differentiate between grayscale and color images.

   A digital image is a two-dimensional grid of pixels represented mathematically as a matrix $F(x,y)$, where each coordinate maps to an an intensity or color value.

   *   **Grayscale Images:** Contain a single channel representing light intensity, typically scaled from 0 (black) to 255 (white) in an 8-bit format.
   *   **Color Images:** Use multiple channels (usually 3) such as Red, Green, and Blue that blend together at each pixel to produce full-spectrum color.

2. Explain the difference between RGB, BGR, HSV, and LAB color spaces.

*   **RGB:** Standard additive color model combining Red, Green, and Blue channels for digital displays and sensors.
*   **BGR:** Uses the same primary channels as RGB but in reverse order; it is the default standard for libraries like OpenCV.
*   **HSV (Hue, Saturation, Value):** Decouples color type (Hue) and vibrancy (Saturation) from brightness (Value), making it ideal for color-based object tracking under changing lighting.
*   **LAB (CIE L*a*b*):** Models human color perception, where $L$ controls lightness and $a/b$ represent color axes (Green–Red and Blue–Yellow), enabling uniform distance-based color comparison.

3. What is the purpose of converting an image to grayscale before further processing?

   Grayscale conversion drops two color channels, reducing image data size by 66% to save computational memory and processing time. Additionally, core computer vision tasks—like Canny edge detection, gradient analysis, and thresholding—depend strictly on intensity changes rather than color information.

4. Explain the concept of image complement (negative image) and mention its practical applications.

   An image negative inverts pixel values using $s = 255 - r$ for an 8-bit image, turning dark areas light and vice versa. It is widely applied in medical imaging (such as X-rays and mammograms) to make faint structures clearer, in astronomical photography to highlight faint celestial bodies, and in film digitization.

5. Differentiate between image resizing, cropping, and scaling.

*   **Resizing:** Changes total image dimensions to a fixed width and height via interpolation, which can distort the original aspect ratio.
*   **Cropping:** Cuts out a specific sub-region matrix from the image and discards the rest, preserving original feature scale inside the crop.
*   **Scaling:** Expands or shrinks the overall image using a uniform factor, ensuring the original aspect ratio is strictly maintained.

6. What is a Region of Interest (ROI), and why is it important in computer vision?

   An ROI is a targeted bounding box or subset of an image selected for dedicated processing. Focusing exclusively on an ROI eliminates unnecessary background computations, speeds up execution time, and prevents background noise from triggering false detections in downstream algorithms.

7. Why is OpenCV preferred over conventional image processing libraries for computer vision applications?

   OpenCV is built in C/C++ with deep hardware optimization (SIMD, OpenCL, and CUDA), making it significantly faster than pure Python libraries like PIL or Matplotlib. It natively supports real-time camera video streams and offers an extensive collection of built-in computer vision algorithms.

8. Explain how image resolution and pixel intensity influence image quality.

*   **Spatial Resolution:** Refers to total pixel count per area. Higher resolution preserves sharper edges and finer structural details, whereas low resolution leads to pixelation.
*   **Pixel Intensity (Bit Depth):** Determines the dynamic range of available shades. Higher bit depths (e.g., 12-bit or 16-bit) yield smoother transitions and prevent detail loss in severe highlights or dark shadows.

9. Mention five real-world applications where basic image preprocessing is an essential step.

*   **Optical Character Recognition (OCR):** Binarizing and straightening document scans before text extraction.
*   **Autonomous Vehicles:** Blurring and perspective warp filtering to isolate lane markers on roads.
*   **Medical Diagnostics:** Histogram equalization to sharpen tissue detail in MRIs and CT scans.
*   **Facial Recognition:** Face alignment, resizing, and illumination normalization prior to feature embedding.
*   **Industrial Inspection:** High-contrast thresholding to detect manufacturing defects on assembly lines.

10. How do image preprocessing techniques improve the performance of feature extraction and deep learning models?

    Preprocessing removes sensor noise, standardizes input tensor shapes required by neural networks, and normalizes pixel distributions to enable faster gradient descent convergence. Enhancing image contrast also ensures that key structural features (like edges and textures) stand out clearly for convolutional filters and feature detectors.
